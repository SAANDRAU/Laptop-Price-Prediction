```python
x=df[['brand', 'processor_brand', 'processor_name', 'processor_gnrtn',
       'ram_gb', 'ram_type', 'ssd', 'hdd', 'os', 'os_bit', 'graphic_card_gb',
       'weight', 'warranty', 'Touchscreen', 'msoffice', 'rating',
       'Number of Ratings']].values
```

```python
y=df[["Price"]].values
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
sc=StandardScaler()
```

```python
x_train=sc.fit_transform(x_train)
```

```python
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor,AdaBoostRegressor,GradientBoostingRegressor
from sklearn.linear_model import Lasso,Ridge
from xgboost import XGBRegressor
```

```python
models={
      "Logistic Regression":LogisticRegression(),
      "Lasso":Lasso(),
      "Ridge":Ridge(),
      "Decision Tree Regressor":DecisionTreeRegressor(),
      "KNN":KNeighborsRegressor(),
      "SVM":SVR(),
      "Random Forest Regressor":RandomForestRegressor(),
      "AdaBoost Regressor":AdaBoostRegressor(),
      "GradientBoostingRegressor":GradientBoostingRegressor(),
      "XGBRegressor":XGBRegressor()
}
```

```python
from sklearn.model_selection import cross_val_score
import warnings
warnings.filterwarnings('ignore')
```

```python
for name,model in models.items():
    scores=cross_val_score(estimator=model,X=x_train,y=y_train,cv=5,scoring="r2")
    print(f"score={np.mean(scores)},model={name}")
    print()
```

```python
param={
      "n_estimators":[100,200,300],
      "learning_rate":[0.01,0.05,0.1],
      "max_depth":[2,3,4],
      "min_samples_split":[2,5,10],
      "min_samples_leaf":[1,2,4]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
gb=GradientBoostingRegressor()
```

```python
grid=GridSearchCV(estimator=gb,param_grid=param,cv=5,scoring="r2")
```

```python
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
gb1=GradientBoostingRegressor(learning_rate= 0.05,
 max_depth= 3,
 min_samples_leaf= 4,
 min_samples_split= 10,
 n_estimators= 200)
```

```python
gb1.fit(x_train,y_train)
```

```python
p=gb1.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mean_absolute_error(y_test,p)
```

```python
r2_score(y_test,p)
```
